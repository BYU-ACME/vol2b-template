```python
class GeneticAlgorithmTSP:
    """
    Genetic algorithm for tours represented by permutations of city indices.

    Parameters
    ----------
    filename : str or path-like
        NPZ file containing ``coordinates`` and ``distance_matrix`` arrays.
    population_size : int
        Number of routes in each generation.
    number_of_cities : int
        Number of cities included in each route.
    seed : int or None, default=None
        Seed used to initialize the NumPy random-number generator.
    """

    def __init__(self, filename, population_size, number_of_cities, seed=None):
        self.locations = filename
        self.population_size = population_size
        self.number_of_cities = number_of_cities
        self.coordinates = None
        self.distance_matrix = None
        self.rng = np.random.default_rng(seed)

    def init_population(self):
        """
        Load the city data and generate random routes.

        Returns
        -------
        population : ndarray of shape (population_size, number_of_cities)
            Random permutations of the city indices.
        """
        # Load the data, extract coordinates and distance matrix
        data = np.load(self.locations)
        self.coordinates = data["coordinates"][:self.number_of_cities]
        self.distance_matrix = data["distance_matrix"][
            :self.number_of_cities, :self.number_of_cities
        ]

        # Generate random routes
        return np.array([
            self.rng.permutation(self.number_of_cities)
            for _ in range(self.population_size)
        ], dtype=int)

    def fitness(self, individual):
        """
        Compute the total distance of one closed route.

        Parameters
        ----------
        individual : ndarray of shape (number_of_cities,)
            Permutation of integer city indices.

        Returns
        -------
        distance : float
            Total route length, including the return to the first city.
        """
        raise NotImplementedError("Problem 5 not implemented")

    def evaluate_population(self, population):
        """
        Compute the shortest and mean route lengths in a population.

        Parameters
        ----------
        population : ndarray of shape (population_size, number_of_cities)
            Routes to evaluate.

        Returns
        -------
        best_route : ndarray of shape (number_of_cities,)
            Copy of the shortest route.
        best_length : float
            Length of the shortest route.
        mean_length : float
            Mean route length in the population.
        """
        raise NotImplementedError("Problem 5 not implemented")

    def selection(self, population, sample_size):
        """
        Select the shortest routes from a population.

        Parameters
        ----------
        population : ndarray of shape (population_size, number_of_cities)
            Population from which to select parents.
        sample_size : int
            Number of routes to include in the parent pool.

        Returns
        -------
        parent_pool : ndarray of shape (sample_size, number_of_cities)
            Copies of the selected routes, ordered from shortest to longest.
        """
        raise NotImplementedError("Problem 5 not implemented")

    def mutation(self, individual):
        """
        Mutate a route by swapping two city positions.

        Parameters
        ----------
        individual : ndarray of shape (number_of_cities,)
            Route to mutate.

        Returns
        -------
        route : ndarray of shape (number_of_cities,)
            Mutated copy of the route.
        """
        raise NotImplementedError("Problem 5 not implemented")

    def crossover(self, parents):
        """
        Combine several parent routes into one valid permutation.

        Parameters
        ----------
        parents : ndarray of shape (num_parents, number_of_cities)
            Parent routes used to construct the child.

        Returns
        -------
        child : ndarray of shape (number_of_cities,)
            Integer route containing every city exactly once.
        """
        raise NotImplementedError("Problem 5 not implemented")

    def genetic_algorithm(self, num_parents=3, sample_size=50, mutation_prob=0.25,
                          max_iters=1000, tol=1e-3, patience=100, plot=False,
                          plot_interval=200, progress=True):
        """
        Evolve a fixed-size population of routes.

        Parameters
        ----------
        num_parents : int, default=3
            Number of distinct parent routes used to create each child.
        sample_size : int, default=50
            Number of selected routes in the parent pool.
        mutation_prob : float, default=0.25
            Probability of mutating each newly created route.
        max_iters : int, default=1000
            Maximum number of generations.
        tol : float, default=1e-3
            Minimum decrease in the best route length counted as improvement.
        patience : int, default=100
            Number of consecutive generations without sufficient improvement
            allowed before stopping.
        plot : bool, default=False
            If True, plot the initial route and selected intermediate routes.
        plot_interval : int, default=200
            Number of generations between intermediate route plots.
        progress : bool, default=True
            If True, display a progress bar.

        Returns
        -------
        route : ndarray of shape (number_of_cities,)
            Shortest route in the final population.
        length : float
            Length of ``route`` in kilometers.
        population : ndarray of shape (population_size, number_of_cities)
            Final population of routes.
        fitness_scores : dict of str to list of float
            Best and mean route lengths for generation 0 and every completed
            generation, stored under ``"best"`` and ``"mean"``.
        """
        iterator = tqdm(range(1, max_iters + 1), disable=not progress)
        raise NotImplementedError("Problem 5 not implemented")

    def plot_route(self, route, title="Final Path"):
        """
        Plot one closed route on a map.

        Parameters
        ----------
        route : ndarray of shape (number_of_cities,)
            Permutation of integer city indices to plot.
        title : str, default="Final Path"
            Title displayed above the map.

        Returns
        -------
        axis : cartopy.mpl.geoaxes.GeoAxes
            Cartopy axis containing the plotted route.
        """
        # Extract coordinates from data file
        longitude, latitude = self.coordinates[route].T

        # Create a closed loop by appending the first point to the end
        longitude = np.append(longitude, longitude[0])
        latitude = np.append(latitude, latitude[0])

        # Create a plot using Cartopy
        projection = ccrs.PlateCarree()
        figure = plt.figure()
        axis = figure.add_subplot(111, projection=projection)
        axis.plot(longitude, latitude, "o-", transform=projection)
        axis.coastlines()
        axis.add_feature(cfeature.STATES)
        axis.gridlines(draw_labels=True)
        axis.set_title(f"{title} (length {self.fitness(route):.1f} km)")
        return axis
```
